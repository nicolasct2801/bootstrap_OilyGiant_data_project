## En este proyecto encontraremos los mejores lugares donde abrir 200 pozos nuevos de petróleo. Creando un modelo que ayude a elegir la región con el mayor margen de beneficio. Analizando los beneficios y riesgos potenciales utilizando la técnica bootstrapping.

se importan las librerias necesarias

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

se cargan los datos y se preprocesan

```python
def cargar_datos(archivo):
    df = pd.read_csv(archivo)
    return df

def preprocesar_datos(df):
    df = df.dropna()
    return df
```

Ahora se entrena el modelo de regresion lineal

```python
def entrenar_modelo(df):
    X = df[['f0', 'f1', 'f2']]
    y = df['product']
    X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.25, random_state=42)
    
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    y_pred = model.predict(X_val)
    
    rmse = mean_squared_error(y_val, y_pred, squared=False)
    print("RMSE:", rmse)
    return y_val, y_pred
```

Ahora creamos una funcion para aplicar la tecnica bootstrapping a las 3 regiones 

```python
def analisis_bootstrapping(df_region, revenue_per_unit=4500, cost_per_well=500000, n_iterations=1000):
    bootstrap_results = []
    
    for _ in range(n_iterations):
        sample_500 = df_region.sample(n=500, replace=True) 
        df_top_200 = sample_500.sort_values(by='pred', ascending=False).head(200)
        df_top_200['real_income'] = df_top_200['real'] * revenue_per_unit  
        df_top_200['real_profit'] = df_top_200['real_income'] - cost_per_well  
        bootstrap_results.append(df_top_200['real_profit'].sum())
    
    bootstrap_results = np.array(bootstrap_results)
    mean_profit = bootstrap_results.mean()
    conf_interval = np.percentile(bootstrap_results, [2.5, 97.5])
    loss_risk = np.mean(bootstrap_results < 0)  
    
    print(f"Beneficio promedio (Bootstrap): {mean_profit}")
    print(f"Intervalo de confianza del 95%: {conf_interval}")
    print(f"Riesgo de pérdida: {loss_risk * 100}%")
    
    return mean_profit, conf_interval, loss_risk
```

definimos 'archivos' que es donde estan los datos para las 3 regiones

```python
archivos = ['/datasets/geo_data_0.csv', '/datasets/geo_data_1.csv', '/datasets/geo_data_2.csv']
```

Ahora analizamos las 3 regiones de extraccion de petroleo utilizando un bucle for

```python
for i, archivo in enumerate(archivos):
    print(f"Analizando región {i}...")
    df = cargar_datos(archivo)
    df = preprocesar_datos(df)
    real, pred = entrenar_modelo(df)
    df_region = pd.DataFrame({'real': real, 'pred': pred})
    analisis_bootstrapping(df_region)
    print("\n------------------------\n")
```
